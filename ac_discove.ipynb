```python
import asyncio
import json
from greeclimate.discovery import Discovery

JSON_FILE = "ac_devices.json"

async def discover_and_save():
    discovery = Discovery()
    devices = await discovery.scan(wait_for=5)

    if not devices:
        print("No devices found!")
        return []

    saved_devices = []
    for dev_info in devices:
        saved_devices.append({
            "mac": dev_info.mac,
            "ip": dev_info.ip
        })
        print(f"Saved AC {dev_info.mac} @ {dev_info.ip}")

    with open(JSON_FILE, "w") as f:
        json.dump(saved_devices, f, indent=2)

    return saved_devices

if __name__ == "__main__":
    asyncio.run(discover_and_save())
```
